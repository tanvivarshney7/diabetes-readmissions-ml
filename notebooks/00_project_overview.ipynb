# Diabetes 130-US Hospitals: 30-Day Readmission Prediction

### End-to-End Data Analysis and Machine Learning Project

This project analyzes hospital encounter data for patients with diabetes and develops machine learning models to predict whether a patient will be readmitted to the hospital within 30 days.

## Project Overview

Hospital readmissions can increase healthcare costs and may indicate that some patients need additional support after discharge.

In this project, I analyze the Diabetes 130-US Hospitals dataset to understand patterns related to hospital readmissions.

The project follows an end-to-end data analytics and machine learning workflow, starting with data understanding and cleaning and ending with model evaluation and business/healthcare recommendations.

The main focus is predicting whether a patient will be readmitted within 30 days of a hospital encounter.

## Business Problem & Project Objective

Hospital readmissions can increase healthcare costs and may indicate that some patients require additional support after discharge.

The main business question for this project is:

> **Can patient and hospital encounter information be used to predict whether a patient will be readmitted within 30 days?**

## Dataset Information

### Dataset Name

Diabetes 130-US Hospitals for Years 1999-2008

### Source

UCI Machine Learning Repository

### Dataset Link

https://archive.ics.uci.edu/dataset/296/diabetes+130-us+hospitals+for+years+1999-2008

### Dataset Description

The dataset contains hospital encounter information for patients with diabetes collected from multiple hospitals in the United States between 1999 and 2008.

The data contains information related to patient demographics, hospital visits, diagnoses, procedures, medications, laboratory tests, and readmission status.

## Dataset Characteristics

The dataset contains information about:

- Patient demographics
- Hospital admission information
- Length of hospital stay
- Laboratory procedures
- Medical procedures
- Medications
- Previous outpatient visits
- Previous emergency visits
- Previous inpatient visits
- Number of diagnoses
- Diabetes-related information
- Readmission status

The dataset contains multiple hospital encounters, and some patients may have more than one encounter.

## Target Variable

The original dataset contains a variable called `readmitted`.

For this project, I created a binary target variable called:

`readmitted_30_days`

### Target Definition

- `0` = Patient was not readmitted within 30 days
- `1` = Patient was readmitted within 30 days

This converts the original readmission information into a binary classification problem.

## Why Predict 30-Day Readmission?

Thirty-day readmission is useful as a healthcare performance measure because it focuses on patients who return to the hospital shortly after their previous encounter.

By focusing on 30-day readmission, the project can explore whether available patient and encounter information can help identify patients who may require additional attention after discharge.

## Models, Tools & Data Preparation

### Machine Learning Models

**Logistic Regression** is used as a baseline classification model because it is simple and easy to interpret.

**Random Forest** is used as a second model because it can capture more complex relationships and provides feature importance information.

### Evaluation

Because the target variable is imbalanced, model performance will not be evaluated using accuracy alone. Precision, recall, F1-score, and ROC-AUC will also be considered.

### Data Leakage Prevention

The dataset contains multiple encounters for some patients. To reduce the risk of the same patient appearing in both training and testing data, the train/test split is performed at the patient level using `patient_nbr`.

The `patient_nbr` variable is used only for grouping patients and is not used as a machine learning feature.

### Project Limitations

The dataset contains historical data from 1999-2008, so healthcare practices may have changed since the data was collected. The analysis identifies patterns and relationships but does not prove causation. The model should be considered a decision-support tool and would require additional validation before real-world healthcare use.

### Final Project Goal

The overall goal is to build an end-to-end data analytics and machine learning project that can identify patterns related to 30-day hospital readmission, compare predictive models, and translate the results into meaningful data-driven insights and recommendations.
